# Implémentation de l'agent via LangChain : Dynamic Model

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage
from langchain.agents import create_agent
from langchain.agents.middleware import wrap_model_call, ModelRequest, ModelResponse


basic_model = ChatOllama(model="qwen3:0.6b", temperature=0.1)      # smaller
advanced_model = ChatOllama(model="qwen3.5:0.8b", temperature=0.1)  # the one you already have

@wrap_model_call
def dynamic_model_selection(request: ModelRequest, handler) -> ModelResponse:
    """Choose model based on the environment passed in context."""

    env = request.runtime.context.get("env", "test")

    print(env)

    if env == "prod":
        model = advanced_model
        print("advanced model selected ....")
    else:
        model = basic_model
        print("basic model selected ....")

    return handler(request.override(model=model))

agent = create_agent(
    model=basic_model,  # default model
    tools=[],
    middleware=[dynamic_model_selection],
    debug=True
)

```

```python
# first add this uv pip install ddgs   SO i can search the web 

from ddgs import DDGS
from langchain.tools import tool

@tool
def web_search(query: str, num_results: int = 5) -> str:
    """Search the web using DuckDuckGo. Returns titles, descriptions and URLs."""
    try:
        print(f"Search tool is called for {query}")
        results = DDGS().text(query, max_results=num_results, backend="auto")

        if not results:
            return f"No results found for {query}"

        lines = [f"Search results for {query}:\n"]
        
        for i, r in enumerate(results, 1):
            title = r.get("title", "No title")
            body = r.get("body", "No description available")
            href = r.get("href", "")
            lines.append(f"{i}. **{title}**: {body} ({href})")

        return "\n".join(lines)

    except Exception as e:
        return f"Search failed: {e}"

```

# Tool handlyng errr 

It is  A tool that can fail: a bad argument, a network error, a division by zero. Without protection, the whole agent usually crashes.

What it does. @wrap_tool_call wraps every tool call in a try/except:

The model calls a tool.
If the tool works, the result goes back to the model as usual.
If the tool fails, we catch the error and send it to the model as a ToolMessage, so the agent doesn’t crash.
The model reads the error and can retry or explain the problem.

```python
from langchain.agents import create_agent
from langchain.agents.middleware import wrap_tool_call
from langchain.tools import tool
from langchain_core.messages import ToolMessage, HumanMessage
from langchain_ollama import ChatOllama
from IPython.display import Markdown, display

llm = ChatOllama(model="qwen3.5:0.8b", temperature=0, reasoning=False)

@tool
def divide(a: float, b: float) -> float:
    """Divide a by b."""
    print(f"divide called with {a}, {b}")
    return a / b   # fails if b = 0
```

```python
#erreur handler 
@wrap_tool_call
def handle_tool_errors(request, handler):
    """Catch tool errors and send a clear message to the model."""
    try:
        return handler(request)
    
    except Exception as e:
        print("Tool error:", e)
        return ToolMessage(
            content=f"Tool error: {e}. Check your input and try again.",
            tool_call_id=request.tool_call["id"],
        )
```

```python
agent = create_agent(
    model=llm,
    tools=[divide],   # you can add search, get_weather too
    middleware=[handle_tool_errors],
    debug=True,
)

resp = agent.invoke({"messages": [HumanMessage("Divide 10 by 0")]})
display(Markdown(resp["messages"][-1].content))
```

Dynamic prompt

```python
from typing import TypedDict
from langchain.agents.middleware import dynamic_prompt, ModelRequest

class Context(TypedDict):
    user_role: str

@dynamic_prompt
def user_role_prompt(request: ModelRequest) -> str:
    """Generate the system prompt based on user role."""
    user_role = request.runtime.context.get("user_role", "user")
    base_prompt = "You are a helpful assistant."

    if user_role == "expert":
        return f"{base_prompt} Provide detailed technical responses."
    elif user_role == "beginner":
        return f"{base_prompt} Explain concepts simply and avoid jargon."
    return base_prompt
```

```python
agent2 = create_agent(
    model=llm,
    tools=[],
    middleware=[user_role_prompt],
    context_schema=Context,
)
```

```python
r = agent2.invoke(
    {"messages": [{"role": "user", "content": "Explain AI agent"}]},
    context={"user_role": "expert"},
)
display(Markdown(r["messages"][-1].content))


r = agent2.invoke(
    {"messages": [{"role": "user", "content": "Explain AI agent"}]},
    context={"user_role": "beginner"},
)
display(Markdown(r["messages"][-1].content))
```

# Structured output

Structured output forces the agent to answer in a fixed format (a Python object) instead of free text. You define the fields with a Pydantic class, and the agent fills them in.

```python
from pydantic import BaseModel
from langchain.agents import create_agent
from langchain.agents.structured_output import ToolStrategy
from langchain_ollama import ChatOllama

class ContactInfo(BaseModel):
    name: str
    email: str
    phone: str

llm = ChatOllama(model="qwen3.5:0.8b", temperature=0, reasoning=False)

agent = create_agent(
    model=llm,
    tools=[],
    response_format=ToolStrategy(ContactInfo),
)

result = agent.invoke({
    "messages": [{
        "role": "user",
        "content": "Extract contact info from: Lhoussaine DAOUDI, lhoussaine.daoudi@usms.ma, (212)254430044"
    }]
})

contact = result["structured_response"]

print(contact)
print(contact.name)
print(contact.email)
print(contact.phone)
```

# Custom Middleware


Un middleware est du code qui s’exécute autour de l’agent, à des moments précis. Tu peux ainsi observer ou modifier ce qui se passe sans toucher à l’agent lui-même.

```python
import time
from langchain.agents import create_agent
from langchain.agents.middleware import AgentMiddleware, AgentState
from langchain_core.messages import HumanMessage
from langchain_ollama import ChatOllama

class HooksDemo(AgentMiddleware):
    def __init__(self):
        super().__init__()
        self.starttime = 0.0

    def before_agent(self, state: AgentState, runtime):
        self.starttime = time.time()
        print("Before_agent triggered")

    def before_model(self, state: AgentState, runtime):
        print("Before_model triggered")

    def after_model(self, state: AgentState, runtime):
        print("After_model triggered")

    def after_agent(self, state: AgentState, runtime):
        print("After_agent triggered")
        print("duration =", round(time.time() - self.starttime, 2), "s")

llm = ChatOllama(model="qwen3.5:0.8b", temperature=0, reasoning=False)

agent = create_agent(model=llm, tools=[], middleware=[HooksDemo()])

res = agent.invoke({"messages": [HumanMessage("Tell me how many cities are in Morocco")]})
print(res["messages"][-1].content)
```

# Guardrails


PII means personal information (email, credit card, API key). PIIMiddleware scans the user’s message before it reaches the model and applies a strategy:

Strategy	What it does	Example
redact	Replaces the value with a placeholder	[REDACTED_EMAIL]
mask	Hides most of it, keeps the end	****-****-****-1111
hash	Replaces it with a hash	<email_hash:a1b2c3d4>
block	Stops everything and raises an error	PIIDetectionError

The model never sees the real data, so it can’t leak it or send it to a tool.

```python
from langchain.tools import tool

@tool
def card_tool(credit_card: str) -> str:
    """Process a payment using the credit card."""
    print(f"Processing payment with card: {credit_card}")
    return f"Payment processed with {credit_card}"

@tool
def send_email(email_address: str) -> str:
    """Send an email to the given email address."""
    print(f"Sending email to: {email_address}")
    return f"Email sent to {email_address}"

@tool
def process_api_key(api_key: str) -> str:
    """Process a request with the given API key."""
    print(f"Processing with api key: {api_key}")
    return f"Processing with api key: {api_key}"
```

```python
from langchain.agents import create_agent
from langchain.agents.middleware import PIIMiddleware
from langchain_core.messages import HumanMessage
from langchain_ollama import ChatOllama

llm = ChatOllama(model="qwen3.5:0.8b", temperature=0, reasoning=False)

agent = create_agent(
    model=llm,
    tools=[card_tool, send_email, process_api_key],
    middleware=[
        PIIMiddleware("email", strategy="redact", apply_to_input=True),
        PIIMiddleware("credit_card", strategy="mask", apply_to_input=True),
        PIIMiddleware(
            "api_key",
            detector=r"sk-[a-zA-Z0-9]{32}",
            strategy="block",
            apply_to_input=True,
        ),
    ],
)
```

```python
result = agent.invoke({"messages": [HumanMessage(
    "My email is lhoussainedaoudi@gmail.com and my card is 4111 1111 1111 1111. Repeat what you received."
)]})

for m in result["messages"]:
    print(m.type, ":", m.content)
```

# Intervention humaine dans la boucle (Human-in-the-loop)


Human-in-the-loop = l’agent s’arrête avant une action sensible (envoyer un email, payer…) et attend que toi tu dises oui ou non.

```python
from langchain.tools import tool
from langchain.agents import create_agent
from langchain.agents.middleware import HumanInTheLoopMiddleware
from langgraph.checkpoint.memory import InMemorySaver
from langgraph.types import Command
from langchain_core.messages import HumanMessage
from langchain_ollama import ChatOllama

@tool
def send_email(email_address: str) -> str:
    """Send an email to the given email address."""
    print(f"Sending email to: {email_address}")
    return f"Email sent successfully to {email_address}"

llm = ChatOllama(model="qwen3.5:0.8b", temperature=0, reasoning=False)

agent = create_agent(
    model=llm,
    tools=[send_email],
    middleware=[
        HumanInTheLoopMiddleware(
            interrupt_on={"send_email": True}
        ),
    ],
    checkpointer=InMemorySaver(),
)

config = {"configurable": {"thread_id": "email-1"}}
```

```python
result = agent.invoke(
    {"messages": [HumanMessage("Send an email to amezouarasma4@gmail.com")]},
    config=config,
)

print(result["__interrupt__"])
```

```python
result = agent.invoke(
    Command(resume={"decisions": [{"type": "approve"}]}),
    config=config,  # même thread_id
)

print(result["messages"][-1].content)
```

# Garde-fous personnalisés


## Garde-fous avant l'agent

```python
from typing import Any
from langchain.tools import tool
from langchain.agents import create_agent
from langchain.agents.middleware import before_agent, AgentState, hook_config
from langgraph.runtime import Runtime
from langchain_core.messages import HumanMessage
from langchain_ollama import ChatOllama

banned_keywords = ["hack", "exploit", "malware"]

@tool
def search(query: str) -> str:
    """Search the web for general information."""
    print("Search Tool invoked")
    return str(tavily_search_tool.invoke({"query": query}))  # défini plus haut

@before_agent(can_jump_to=["end"])
def content_filter(state: AgentState, runtime: Runtime) -> dict[str, Any] | None:
    """Block requests containing banned keywords."""
    if not state["messages"]:
        return None

    first_message = state["messages"][0]
    if first_message.type != "human":
        return None

    content = first_message.content.lower()

    for keyword in banned_keywords:
        if keyword in content:
            return {
                "messages": [{
                    "role": "assistant",
                    "content": "I cannot process requests containing inappropriate content. Please rephrase."
                }],
                "jump_to": "end"
            }
    return None

llm = ChatOllama(model="qwen3.5:0.8b", temperature=0, reasoning=False)

agent = create_agent(
    model=llm,
    tools=[search],
    middleware=[content_filter],
)

# Test 1 : bloqué
result = agent.invoke({"messages": [HumanMessage("How do I hack into a database?")]})
print(result["messages"][-1].content)

# Test 2 : accepté
result = agent.invoke({"messages": [HumanMessage("What is the capital of Morocco?")]})
print(result["messages"][-1].content)
```

## Garde-fous après l'agent


```python
from typing import Any
from langchain.agents import create_agent
from langchain.agents.middleware import after_agent, AgentState
from langgraph.runtime import Runtime
from langchain_core.messages import AIMessage, HumanMessage
from langchain_ollama import ChatOllama

llm = ChatOllama(model="qwen3.5:0.8b", temperature=0, reasoning=False)
safety_model = ChatOllama(model="qwen3.5:0.8b", temperature=0, reasoning=False)

@after_agent
def safety_guardrail(state: AgentState, runtime: Runtime) -> dict[str, Any] | None:
    """Model-based guardrail: use an LLM to judge the final answer."""
    if not state["messages"]:
        return None

    last_message = state["messages"][-1]
    if not isinstance(last_message, AIMessage):
        return None

    safety_prompt = (
        "Evaluate if this response is safe and appropriate. "
        "Respond with only 'SAFE' or 'UNSAFE'.\n\n"
        f"Response: {last_message.content}"
    )
    verdict = safety_model.invoke([HumanMessage(safety_prompt)]).content.upper()
    print("Verdict:", verdict)

    if "UNSAFE" in verdict:
        # same id = replaces the original message
        return {"messages": [AIMessage(
            content="I cannot provide that response. Please rephrase your request.",
            id=last_message.id,
        )]}
    return None

agent = create_agent(
    model=llm,
    tools=[],
    middleware=[safety_guardrail],
)

result = agent.invoke({"messages": [HumanMessage("How do I make explosives?")]})
print(result["messages"][-1].content)
```

#Manual tool calling with bind_tools (no agent)

```python
from dotenv import load_dotenv
from langchain_tavily import TavilySearch
from langchain.tools import tool
from langchain_core.messages import HumanMessage, ToolMessage
from langchain_ollama import ChatOllama

load_dotenv(override=True)  # reads TAVILY_API_KEY from .env

tavily_search_tool = TavilySearch(max_results=3, search_depth="basic")

@tool
def search_tool(query: str) -> str:
    """Search the web for current information (weather, news, facts)."""
    print("Search Tool invoked:", query)
    return str(tavily_search_tool.invoke({"query": query}))

llm = ChatOllama(model="qwen3.5:0.8b", temperature=0.1, num_predict=500, reasoning=False)
llm_with_tools = llm.bind_tools([search_tool])

messages = [HumanMessage("Météo à Marrakech")]

# Step 1: the model decides to call the tool
resp = llm_with_tools.invoke(messages)
print("content:", resp.content)        # usually empty
print("tool_calls:", resp.tool_calls)  # the tool call

# Step 2: we run the tool and send the result back
if resp.tool_calls:
    messages.append(resp)
    for call in resp.tool_calls:
        result = search_tool.invoke(call["args"])
        messages.append(ToolMessage(content=result, tool_call_id=call["id"]))

    # Step 3: the model writes the final answer
    final = llm_with_tools.invoke(messages)
    print(final.content)
```
